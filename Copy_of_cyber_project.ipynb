```python
import os
import re
import hashlib
from datetime import datetime



# ==========================================================
#                    SENTINEL AI
#             CYBER SECURITY MINI PROJECT
# ==========================================================


# ----------------------------------------------------------
# 1. PASSWORD HASHING
# ----------------------------------------------------------

def hash_password(password):
    return hashlib.sha256(password.encode()).hexdigest()


# ----------------------------------------------------------
# 2. ACTIVITY LOG
# ----------------------------------------------------------

def log_activity(activity):

    with open("security_log.txt", "a") as file:

        time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

        file.write(time + " - " + activity + "\n")


# ----------------------------------------------------------
# 3. USER REGISTRATION
# ----------------------------------------------------------

def register():

    print("\n" + "=" * 50)
    print("              USER REGISTRATION")
    print("=" * 50)

    username = input("Enter username: ")
    password = input("Enter password: ")

    if len(password) < 8:

        print("\nPassword must contain at least 8 characters.")
        return

    if os.path.exists("users.txt"):

        with open("users.txt", "r") as file:

            users = file.readlines()

        for user in users:

            if "|" in user:

                saved_username = user.strip().split("|")[0]

                if saved_username == username:

                    print("\nUsername already exists.")
                    return

    hashed_password = hash_password(password)

    with open("users.txt", "a") as file:

        file.write(
            username + "|" + hashed_password + "\n"
        )

    print("\nRegistration successful!")

    log_activity(
        "New user registered: " + username
    )


# ----------------------------------------------------------
# 4. USER LOGIN
# ----------------------------------------------------------

def login():

    print("\n" + "=" * 50)
    print("                  USER LOGIN")
    print("=" * 50)

    username = input("Enter username: ")
    password = input("Enter password: ")

    if not os.path.exists("users.txt"):

        print("\nNo registered users found.")
        return False

    hashed_password = hash_password(password)

    with open("users.txt", "r") as file:

        users = file.readlines()

    for user in users:

        if "|" in user:

            saved_username, saved_password = user.strip().split("|")

            if (
                username == saved_username
                and hashed_password == saved_password
            ):

                print("\nLogin successful!")

                log_activity(
                    "User logged in: " + username
                )

                return True

    print("\nInvalid username or password.")

    log_activity(
        "Failed login attempt: " + username
    )

    return False


# ----------------------------------------------------------
# 5. PASSWORD SECURITY CHECK
# ----------------------------------------------------------

def password_security():

    print("\n" + "=" * 50)
    print("            PASSWORD SECURITY CHECK")
    print("=" * 50)

    password = input("Enter password to check: ")

    score = 0

    if len(password) >= 8:

        print("Length: Good")
        score += 1

    else:

        print("Length: Too short")

    if re.search(r"[A-Z]", password):

        print("Uppercase letter: Present")
        score += 1

    else:

        print("Uppercase letter: Missing")

    if re.search(r"[a-z]", password):

        print("Lowercase letter: Present")
        score += 1

    else:

        print("Lowercase letter: Missing")

    if re.search(r"[0-9]", password):

        print("Number: Present")
        score += 1

    else:

        print("Number: Missing")

    if re.search(r"[@#$%^&*!]", password):

        print("Special character: Present")
        score += 1

    else:

        print("Special character: Missing")

    print("\nSecurity Score:", score, "/ 5")

    if score == 5:

        print("Password Strength: STRONG")

    elif score >= 3:

        print("Password Strength: MEDIUM")

    else:

        print("Password Strength: WEAK")

    log_activity(
        "Password security checked"
    )


# ----------------------------------------------------------
# 6. URL SECURITY CHECK
# ----------------------------------------------------------


def url_security():

    print("\n" + "=" * 50)
    print("             URL SECURITY CHECK")
    print("=" * 50)

    url = input("Enter URL: ").strip()

    # Add https:// if user enters only google.com
    check_url = url
    if not check_url.lower().startswith(("http://", "https://")):
        check_url = "https://" + check_url

    # ---------------- URL VALIDATION ----------------
    try:
        parsed = urlparse(check_url)
        hostname = parsed.hostname

        # Check basic URL format
        if (
            parsed.scheme not in ["http", "https"]
            or not hostname
            or " " in url
            or "." not in hostname
            or not re.match(r"^[A-Za-z0-9.-]+$", hostname)
            or hostname.startswith(".")
            or hostname.endswith(".")
        ):
            print("\nURL Risk Score: 10")
            print("URL Status: INVALID / HIGH RISK")
            print("Reason: Invalid URL format.")
            input("\nPress Enter to continue...")
            return

    except Exception:
        print("\nURL Risk Score: 10")
        print("URL Status: INVALID / HIGH RISK")
        print("Reason: Invalid URL format.")
        input("\nPress Enter to continue...")
        return

    # ---------------- RISK CHECKING ----------------
    risk_score = 0

    suspicious_words = [
        "login",
        "verify",
        "password",
        "account",
        "bank",
        "free",
        "confirm",
        "update",
        "winner",
        "claim"
    ]

    # Suspicious keyword check
    for word in suspicious_words:
        if word in check_url.lower():
            risk_score += 1
            print("Suspicious keyword detected:", word)

    # HTTP check
    if check_url.lower().startswith("http://"):
        risk_score += 1
        print("Warning: HTTP connection detected.")

    # @ symbol check
    if "@" in check_url:
        risk_score += 2
        print("Warning: Suspicious @ symbol detected.")

    # Long URL check
    if len(check_url) > 75:
        risk_score += 1
        print("Warning: Unusually long URL.")

    # ---------------- FINAL RESULT ----------------
    print("\nURL Risk Score:", risk_score)

    if risk_score >= 4:
        print("URL Status: HIGH RISK")

    elif risk_score >= 2:
        print("URL Status: SUSPICIOUS")

    else:
        print("URL Status: LOW RISK / SAFE")

    input("\nPress Enter to continue...")








# ----------------------------------------------------------
# 7. FILE SECURITY CHECK
# ----------------------------------------------------------

def file_security():

    print("\n" + "=" * 50)
    print("               FILE SECURITY CHECK")
    print("=" * 50)

    filename = input("Enter file name: ")

    if not os.path.exists(filename):

        print("\nFile not found.")

        log_activity(
            "File scan failed - File not found"
        )

        return

    extension = os.path.splitext(
        filename
    )[1].lower()

    dangerous_extensions = [
        ".exe",
        ".bat",
        ".cmd",
        ".vbs",
        ".scr",
        ".ps1"
    ]

    print("\nFile Name:", filename)
    print("File Type:", extension)

    if extension in dangerous_extensions:

        print(
            "File Status: POTENTIALLY DANGEROUS"
        )

        log_activity(
            "Dangerous file detected: "
            + filename
        )

    else:

        print(
            "File Status: APPEARS SAFE"
        )

        log_activity(
            "File scanned: "
            + filename
        )


# ----------------------------------------------------------
# 8. EMAIL SECURITY CHECK
# ----------------------------------------------------------

def email_security():

    print("\n" + "=" * 50)
    print("               EMAIL SECURITY CHECK")
    print("=" * 50)

    email = input("Enter email address: ")

    risk_score = 0

    # Basic email format check
    email_pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"

    if re.match(email_pattern, email):

        print("Email Format: VALID")

    else:

        print("Email Format: INVALID")

        risk_score += 2

    # Suspicious words
    suspicious_words = [
        "winner",
        "prize",
        "free",
        "urgent",
        "verify",
        "claim",
        "password",
        "bank",
        "security"
    ]

    for word in suspicious_words:

        if word in email.lower():

            risk_score += 1

            print(
                "Suspicious word detected:",
                word
            )

    # Suspicious temporary/disposable-style domains
    suspicious_domains = [
        "tempmail",
        "10minutemail",
        "guerrillamail",
        "mailinator",
        "yopmail"
    ]

    for domain in suspicious_domains:

        if domain in email.lower():

            risk_score += 2

            print(
                "Suspicious email domain detected."
            )

    # Check multiple @ symbols
    if email.count("@") != 1:

        risk_score += 2

        print(
            "Invalid number of @ symbols."
        )

    print("\nEmail Risk Score:", risk_score)

    if risk_score >= 4:

        print("Email Status: HIGH RISK")

    elif risk_score >= 2:

        print("Email Status: SUSPICIOUS")

    else:

        print("Email Status: APPEARS SAFE")

    log_activity(
        "Email checked - Risk Score: "
        + str(risk_score)
    )


# ----------------------------------------------------------
# 9. VIEW SECURITY LOG
# ----------------------------------------------------------

def view_logs():

    print("\n" + "=" * 50)
    print("                SECURITY LOG")
    print("=" * 50)

    if not os.path.exists(
        "security_log.txt"
    ):

        print("No security activity found.")
        return

    with open(
        "security_log.txt",
        "r"
    ) as file:

        logs = file.read()

    if logs.strip() == "":

        print("No security activity found.")

    else:

        print(logs)


# ----------------------------------------------------------
# 10. SECURITY MENU
# ----------------------------------------------------------

def security_menu():

    while True:

        print("\n" + "=" * 50)
        print("              SENTINEL AI")
        print("          SECURITY DASHBOARD")
        print("=" * 50)

        print("\n1. Password Security Check")
        print("2. URL Security Check")
        print("3. File Security Check")
        print("4. Email Security Check")
        print("5. View Security Log")
        print("6. Logout")

        choice = input(
            "\nEnter your choice: "
        )

        if choice == "1":

            password_security()

        elif choice == "2":

            url_security()

        elif choice == "3":

            file_security()

        elif choice == "4":

            email_security()

        elif choice == "5":

            view_logs()

        elif choice == "6":

            print("\nLogged out successfully.")

            log_activity(
                "User logged out"
            )

            break

        else:

            print("\nInvalid choice.")

        input(
            "\nPress Enter to continue..."
        )


# ----------------------------------------------------------
# 11. MAIN PROGRAM
# ----------------------------------------------------------

def main():

    while True:

        print("\n" + "=" * 55)
        print("                 SENTINEL AI")
        print("          CYBER SECURITY SYSTEM")
        print("=" * 55)

        print("\n1. Register")
        print("2. Login")
        print("3. Exit")

        choice = input(
            "\nEnter your choice: "
        )

        if choice == "1":

            register()

        elif choice == "2":

            if login():

                security_menu()

        elif choice == "3":

            print(
                "\nThank you for using Sentinel AI."
            )

            print(
                "Stay Safe. Stay Secure!"
            )

            break

        else:

            print("\nInvalid choice.")

        input(
            "\nPress Enter to continue..."
        )



# PROGRAM START
if __name__ == "__main__":

  +V
    main()
```
